## 4. Financial Ratio Calculations

```python
df["Revenue Growth (%)"] = df["Revenue from Operations"].pct_change() * 100
```

```python
df[["Financial Year", "Revenue from Operations", "Revenue Growth (%)"]]
```

```python
df["Net Profit Margin (%)"] = (
    df["Profit After Tax"]/df["Revenue from Operations"]
) * 100
```

```python
df[["Financial Year", "Revenue from Operations", "Profit After Tax", "Net Profit Margin (%)"]]
```

```python
df["Debt-to-Equity Ratio"] = df["Total Debt"] / df["Total Equity"]
```

```python
df[["Financial Year", "Total Debt", "Total Equity", "Debt-to-Equity Ratio"]]
```

```python
df["Current Ratio"] = df["Current Assets"] / df["Current Liabilities"]
```

```python
df[["Financial Year", "Current Assets", "Current Liabilities", "Current Ratio"]]
```

```python
df["Quick Ratio"] = (df["Current Assets"] - df["Inventory"])/df["Current Liabilities"]
```

```python
df[["Financial Year", "Inventory", "Quick Ratio"]]
```

```python
opening_assets = 20255.87

df["Average Total Assets"] = (df["Total Assets"].shift(1) + df["Total Assets"])/2

df.loc[0, "Average Total Assets"] = (opening_assets + df.loc[0, "Total Assets"])/2
```

```python
df[["Financial Year", "Total Assets", "Average Total Assets"]]
```

```python
df["ROA (%)"] = (df["Profit After Tax"]/df["Average Total Assets"])*100
```

```python
df[["Financial Year", "Profit After Tax", "Average Total Assets", "ROA (%)"]]
```

```python
opening_equity = 15627.04

df["Average Total Equity"] = (df["Total Equity"].shift(1) + df["Total Equity"])/2

df.loc[0, "Average Total Equity"] = (opening_equity + df.loc[0, "Total Equity"])/2
```

```python
df[["Financial Year", "Total Equity", "Average Total Equity"]]
```

```python
df["ROE (%)"]  = (df["Profit After Tax"] / df["Average Total Equity"])*100
```

```python
df[["Financial Year", "Profit After Tax", "Average Total Equity", "ROE (%)"]]
```

```python
df["EBIT"] = df["Profit Before Tax"] + df["Finance Cost"]
```

```python
df[["Financial Year", "Profit Before Tax", "Finance Cost", "EBIT"]]
```

```python
df["Interest Coverage Ratio"] = (df["EBIT"]/df["Finance Cost"])
```

```python
df[["Financial Year", "EBIT", "Finance Cost", "Interest Coverage Ratio"]]
```

```python
df["Asset Turnover Ratio"] = (df["Revenue from Operations"] / df["Average Total Assets"])
```

```python
df[["Financial Year", "Revenue from Operations", "Average Total Assets", "Asset Turnover Ratio"]]
```

## 5. Financial Performance Visualizations

### 5.1 Revenue Growth

```python
plt.figure(figsize=(8, 5))

plt.plot(
    df["Financial Year"],
    df["Revenue Growth (%)"],
    marker="o",
    color="teal",
    linewidth=2,
    markersize=7
)

plt.title("Year-on-Year Revenue Growth")
plt.xlabel("Financial Year")
plt.ylabel("Revenue Growth (%)")

plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()

plt.savefig(
    r"C:\Users\mansi\OneDrive\Documents\GitHub\Financial Performance & Ratio Analysis\Images\revenue_growth.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

**Interpretation:** Revenue increased consistently throughout the five-year period. Year-on-year growth was highest in FY 2024–25 at 15.87%, followed by 15.17% in FY 2025–26. This indicates that revenue continued to grow, although the growth rate decreased slightly in the latest year.

## 6. Key Insights

* **Revenue Growth:** Revenue increased every year, rising from ₹12,425.37 crore in FY 2021–22 to ₹20,290.43 crore in FY 2025–26. Annual revenue growth was strongest in FY 2024–25 at 15.87%.

* **Net Profit Margin:** Net profit margin peaked at 9.58% in FY 2022–23, declined to 7.31% in FY 2024–25, and recovered slightly to 7.62% in FY 2025–26.

* **Debt-to-Equity Ratio:** The ratio peaked at 0.17 in FY 2023–24 before falling to 0.09 in FY 2025–26, indicating relatively low debt compared with equity in the latest year.

* **Liquidity:** The current ratio declined to 1.20 in FY 2023–24 and recovered to 1.57 by FY 2025–26. The quick ratio also improved, reaching 1.03 in FY 2025–26.

* **Return on Assets (ROA):** ROA peaked at 6.01% in FY 2022–23, declined to 4.29% in FY 2024–25, and recovered to 4.66% in FY 2025–26.

* **Return on Equity (ROE):** ROE reached its highest level of 7.90% in FY 2022–23, declined over the following two years, and improved slightly to 6.94% in FY 2025–26.

* **Interest Coverage Ratio:** The ratio fell from 21.58 times in FY 2022–23 to 7.12 times in FY 2024–25, before recovering to 16.86 times in FY 2025–26.

* **Asset Turnover Ratio:** Asset turnover remained relatively stable, ranging from approximately 0.59 to 0.63 times, and recovered to 0.61 times in FY 2025–26.


## 7. Conclusion
The financial analysis of Tata Consumer Products Limited from FY 2021–22 to FY 2025–26 shows consistent revenue growth, with revenue increasing from ₹12,425.37 crore to ₹20,290.43 crore. However, profitability margins and returns on assets and equity fluctuated during the period. The company's liquidity ratios improved after declining in FY 2023–24, while its debt-to-equity ratio remained relatively low. The interest coverage ratio also recovered in FY 2025–26 after declining in the previous years. Overall, the company demonstrated revenue growth and an improvement in several financial indicators in the latest year, although profitability and asset efficiency remain important areas to monitor.
